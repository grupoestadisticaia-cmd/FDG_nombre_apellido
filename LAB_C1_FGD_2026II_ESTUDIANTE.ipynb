## Actividad 4 — Caso El Buen Sabor: estructura de la base de datos SQLite
**3 puntos · Celdas 1, 2 y 3**

**Caso: El Buen Sabor.** Usted es analista de datos de El Buen Sabor, un restaurante con sedes en Lima, Cusco y Juliaca. La administradora le entrega una base pequeña con 5 ventas simuladas y le pide evaluar si conviene lanzar una promoción. Antes de consultar nada, debe entender cómo está organizada la base de datos. Las cifras son didácticas y no describen una tendencia real.

**Instrucciones:** ejecute las celdas 1 y 2, que crean y llenan la base de datos SQLite con tres tablas: `sedes`, `productos` y `ventas`. Luego ejecute la celda 3, que muestra el contenido de cada tabla, y complete las tablas 4.1 a 4.3 leyendo la estructura en el código de la celda 1. En SQLite un valor booleano se guarda como entero: 1 (sí) o 0 (no).

**Celda 1 de 7**

```python
import sqlite3

conexion = sqlite3.connect(":memory:")
conexion.executescript("""
CREATE TABLE sedes (
    sede_id   INTEGER PRIMARY KEY,
    ciudad    TEXT NOT NULL,
    es_propia INTEGER NOT NULL   -- 1 = sí, 0 = no
);
CREATE TABLE productos (
    producto_id  INTEGER PRIMARY KEY,
    nombre       TEXT NOT NULL,
    categoria    TEXT NOT NULL,
    precio_lista REAL NOT NULL
);
CREATE TABLE ventas (
    venta_id     INTEGER PRIMARY KEY,
    ciudad       TEXT NOT NULL,
    producto     TEXT NOT NULL,
    unidades     INTEGER NOT NULL,
    monto        REAL NOT NULL,
    satisfaccion TEXT            -- Baja, Media o Alta (puede quedar vacía)
);
""")
```

**Celda 2 de 7**

```python
conexion.executescript("""
INSERT INTO sedes VALUES
    (1, 'Lima', 1),
    (2, 'Cusco', 0),
    (3, 'Juliaca', 0);
INSERT INTO productos VALUES
    (1, 'Lomo saltado', 'Plato', 38.0),
    (2, 'Ceviche', 'Plato', 42.0),
    (3, 'Chicha morada', 'Bebida', 9.0),
    (4, 'Limonada', 'Bebida', 8.0),
    (5, 'Suspiro limeño', 'Postre', 15.0);
INSERT INTO ventas VALUES
    (1, 'Lima', 'Lomo saltado', 2, 76.0, 'Alta'),
    (2, 'Cusco', 'Ceviche', 3, 126.0, 'Media'),
    (3, 'Juliaca', 'Chicha morada', 2, 18.0, 'Baja'),
    (4, 'Lima', 'Limonada', 3, 24.0, NULL),
    (5, 'Juliaca', 'Lomo saltado', 1, 38.0, 'Alta');
""")
print("Base de datos SQLite lista.")
```

**Celda 3 de 7**

```python
consulta = """SELECT * FROM sedes;"""
print(conexion.execute(consulta).fetchall())

consulta = """SELECT * FROM productos;"""
print(conexion.execute(consulta).fetchall())

consulta = """SELECT * FROM ventas;"""
print(conexion.execute(consulta).fetchall())
```

### Consulta 1 — función COUNT · 1 punto

**Pregunta del analista:** ¿Qué ventas fueron de S/ 50 o más, de mayor a menor, y cuántas son?

**Celda 4 de 7**

```python
import pandas as pd
consulta = """SELECT ciudad, producto, monto
FROM ventas
WHERE monto >= 50
ORDER BY monto DESC;"""
resultado= pd.read_sql_query(consulta, conexion)
display(resultado)


consulta = """SELECT COUNT(*) AS ventas_de_50_o_mas
FROM ventas
WHERE monto >= 50;"""
resultado= pd.read_sql_query(consulta, conexion)
display(resultado)
```

**Pregunta del analista:** ¿Cuáles son las dos ventas de menor monto y cuál es el monto mínimo registrado?

**Celda 5 de 7**

```python
consulta = """SELECT ciudad, producto, monto
FROM ventas
ORDER BY monto ASC
LIMIT 2;"""
print(conexion.execute(consulta).fetchall())

consulta = """SELECT MIN(monto) AS venta_minima
FROM ventas;"""
print(conexion.execute(consulta).fetchall())
```

### Consulta 4 — función AVG · 1 punto

```python
consulta = """SELECT nombre, precio_lista
FROM productos
WHERE categoria = 'Plato';"""
print(conexion.execute(consulta).fetchall())

consulta = """SELECT AVG(precio_lista) AS precio_promedio_plato
FROM productos
WHERE categoria = 'Plato';"""
print(conexion.execute(consulta).fetchall())
```
